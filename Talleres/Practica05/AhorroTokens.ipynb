# Directivas de Ahorro de Tokens (MCP) .windsurfrules

## Uso de Playwright
- NUNCA tomes capturas de pantalla (screenshots) ni pidas el DOM completo/HTML de la página.
- Extrae únicamente el texto visible o el selector CSS puntual del elemento que se está testeando.
- Para verificar APIs o respuestas JSON, usa peticiones HTTP directas en lugar de abrir el navegador.

## Uso de Memory
- Guarda en memoria persistente ÚNICAMENTE:
  1. Causa raíz y solución final de errores de configuración o entorno recurrentes.
  2. Puertos locales y variables clave del proyecto.
- PROHIBIDO guardar código fuente completo, trazas de error (stack traces) o historial de chat en la memoria.

## Manejo de Errores y Pruebas
- Si una prueba falla, ejecuta SOLO ese test unitario individual, nunca la suite completa.
- Si el mismo error persiste tras 2 intentos de solución, DETENTE y explica la hipótesis conceptual antes de editar más archivos.


Actúa como un arquitecto de software senior y especialista en optimización de contexto para agentes de IA.

Quiero implementar una estrategia de bajo consumo de tokens en este espacio de trabajo para evitar escaneos globales del código. Para ello, realiza las siguientes 3 tareas:

1. CREAR `PROJECT_MAP.md` (en la raíz):
- Analiza la estructura general de las carpetas (Unidad1_Frontend, Unidad2_Backend) y los archivos markdown existentes de planificación/sílabo.
- Genera un resumen arquitectónico ultra-compacto (máximo 80-100 líneas) que describa la tecnología de cada parte, la responsabilidad de cada carpeta y la ubicación de los archivos clave. No incluyas código extenso, solo rutas y descripciones de una línea.

2. CREAR `REQUIREMENTS_CHECKLIST.md` (en la raíz):
- A partir de los documentos `Guia_Practica_Tema5_leccion.md`, `Planificacion_Unidad1_Tema5.md` y `Silabo_Framework_Programacion_Web.md`, extrae y consolida todos los requerimientos funcionales y técnicos del proyecto.
- Organízalos como una lista de verificación con casillas markdown (`- [ ] Requerimiento`), agrupados por hitos, frontend y backend.
- Incluye un apartado inicial con el "Estado General" (Hito actual, Última tarea completada, Siguiente tarea pendiente).

3. CREAR/ACTUALIZAR `.windsurfrules` (en la raíz):
- Configura las reglas del agente con las siguientes instrucciones estrictas:
  a) "Navegación: NUNCA hagas escaneos recursivos ni leas carpetas completas. Consulta primero `PROJECT_MAP.md` para ubicar el archivo exacto antes de leer código."
  b) "Seguimiento de tareas: Al iniciar cualquier sesión o tarea, lee PRIMERO `REQUIREMENTS_CHECKLIST.md` para conocer el estado y qué falta por implementar. No explores el código para adivinar el progreso."
  c) "Actualización continua: Cada vez que completes una tarea, prueba o endpoint con éxito, debes actualizar inmediatamente `REQUIREMENTS_CHECKLIST.md` marcando la casilla correspondiente (`- [x]`) y actualizando la sección de estado."
  d) "Ahorro de tokens: Lee únicamente los archivos que vas a modificar de forma puntual. Prohibido reescribir archivos enteros si solo cambia una función."
